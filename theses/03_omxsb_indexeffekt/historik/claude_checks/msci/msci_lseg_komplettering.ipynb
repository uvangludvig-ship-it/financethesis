```python
import pandas as pd
try:
    import lseg.data as ld
except Exception:
    import refinitiv.data as ld
ld.open_session()
print('session öppen')
```

```python
RICS = ["COREa.ST", "ELUXa.ST", "MTGa.ST", "VICPA.ST^F19", "HUFVc.ST^A20", "IFSa.ST^J16"]
rader = []
for r in RICS:
    try:
        h = ld.get_history(universe=r, start='2006-01-01', end='2026-10-31', interval='daily')
        if h is not None and len(h) > 0:
            h = h.reset_index(); h.columns = ['Date'] + [str(c) for c in h.columns[1:]]; h['ric'] = r; rader.append(h)
            print(r, len(h), 'dagar', h.Date.iloc[0], 'till', h.Date.iloc[-1], '| kolumner:', list(h.columns)[:6])
        else:
            print(r, 'inga data')
    except Exception as e:
        print(r, 'FEL', str(e)[:100])
P = pd.concat(rader, ignore_index=True)
P.to_csv('msci_tvilling_priser_2.csv', index=False)
print('Klart:', P.ric.nunique(), 'av', len(RICS), '. Ladda upp msci_tvilling_priser_2.csv till Claude.')
```
