# Tvillingaktier: hämta fyra A-aktier från LSEG
Kör cellerna i ordning med **Shift + Enter**. Ladda sedan ner `tvilling_priser_lseg.csv` (högerklicka i filpanelen → Download) och ladda upp den till Claude.

```python
import pandas as pd
try:
    import lseg.data as ld
except Exception:
    import refinitiv.data as ld
ld.open_session()
print('session öppen')
```

```python
SAKNAS = ['NCCa.ST', 'RATOa.ST', 'SWECa.ST', 'BONAVa.ST']
FIELDS = ['TR.PriceClose', 'TR.Volume', 'TR.TotalReturn1D']
out = []
for r in SAKNAS:
    try:
        h = ld.get_history(universe=r, fields=FIELDS, start='2006-01-01', end='2026-10-31', interval='daily')
        h = h.reset_index(); h['ric'] = r
        out.append(h)
        print(r, 'OK', len(h), 'dagar, från', h.iloc[0, 0], 'till', h.iloc[-1, 0])
    except Exception as e:
        print(r, 'FEL', str(e)[:150])
```

```python
df = pd.concat(out, ignore_index=True)
df.to_csv('tvilling_priser_lseg.csv', index=False)
print('sparat:', len(df), 'rader')
df.head()
```
